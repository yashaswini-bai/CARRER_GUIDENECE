```python
# ============================================================
# CAREERCOMPASS AI - JUPYTER NOTEBOOK VERSION
# AI-Based Career and Skill Gap Analyzer
# Fixed 3D Robotic Dashboard
# ============================================================

import tkinter as tk
from tkinter import ttk, messagebox
import random
import math

# ============================================================
# CAREER DATABASE
# ============================================================

CAREER_DATA = {
    "Software Developer": {
        "skills": ["Python", "Java", "SQL", "Git",
                   "Data Structures", "OOP", "Problem Solving"],
        "courses": ["Python Programming", "Java Programming",
                    "SQL Database", "Data Structures",
                    "Object Oriented Programming", "Git & GitHub"]
    },

    "Data Analyst": {
        "skills": ["Python", "SQL", "Excel", "Statistics",
                   "Power BI", "Data Visualization"],
        "courses": ["Python for Data Analysis", "SQL",
                    "Advanced Excel", "Statistics",
                    "Power BI", "Data Visualization"]
    },

    "AI / ML Engineer": {
        "skills": ["Python", "Machine Learning", "SQL",
                   "Mathematics", "Statistics", "Pandas", "NumPy"],
        "courses": ["Python", "Machine Learning",
                    "Statistics", "Mathematics for AI",
                    "Pandas", "NumPy"]
    },

    "Web Developer": {
        "skills": ["HTML", "CSS", "JavaScript",
                   "React", "Git", "Bootstrap"],
        "courses": ["HTML", "CSS", "JavaScript",
                    "React", "Bootstrap", "Git & GitHub"]
    },

    "Cyber Security Analyst": {
        "skills": ["Networking", "Linux", "Python",
                   "Cyber Security", "Cryptography", "Ethical Hacking"],
        "courses": ["Computer Networking", "Linux",
                    "Python for Cyber Security",
                    "Cyber Security Fundamentals",
                    "Cryptography", "Ethical Hacking"]
    },

    "Cloud Engineer": {
        "skills": ["Linux", "Networking", "Python",
                   "AWS", "Docker", "Git"],
        "courses": ["Linux", "Computer Networking",
                    "Python", "AWS Cloud",
                    "Docker", "Git & GitHub"]
    }
}

# ============================================================
# COLORS
# ============================================================

BG = "#080B18"
PANEL = "#10152A"
PANEL2 = "#151B35"
HEADER = "#0B1022"

CYAN = "#00F5FF"
PURPLE = "#9D4EDD"
PINK = "#FF4ECD"
GREEN = "#00E5A8"
RED = "#FF5470"
YELLOW = "#FFD166"

WHITE = "#F5F7FF"
TEXT = "#C9D1E8"
MUTED = "#77819D"

# ============================================================
# MAIN WINDOW
# ============================================================

root = tk.Tk()
root.title("CareerCompass AI - 3D Robotic Dashboard")
root.geometry("1250x750")
root.minsize(1050, 650)
root.configure(bg=BG)

# ============================================================
# HEADER
# ============================================================

header = tk.Frame(root, bg=HEADER, height=75)
header.pack(fill="x")
header.pack_propagate(False)

title_frame = tk.Frame(header, bg=HEADER)
title_frame.pack(side="left", padx=25, pady=10)

tk.Label(
    title_frame,
    text="CAREERCOMPASS AI",
    font=("Arial", 24, "bold"),
    fg=CYAN,
    bg=HEADER
).pack(anchor="w")

tk.Label(
    title_frame,
    text="AI-BASED CAREER & SKILL GAP ANALYZER",
    font=("Arial", 9, "bold"),
    fg=MUTED,
    bg=HEADER
).pack(anchor="w")

tk.Label(
    header,
    text="● AI SYSTEM ONLINE",
    font=("Arial", 10, "bold"),
    fg=GREEN,
    bg=HEADER
).pack(side="right", padx=30)

# ============================================================
# BODY
# ============================================================

body = tk.Frame(root, bg=BG)
body.pack(fill="both", expand=True)

# ============================================================
# SIDEBAR
# ============================================================

sidebar = tk.Frame(body, bg=HEADER, width=220)
sidebar.pack(side="left", fill="y")
sidebar.pack_propagate(False)

tk.Label(
    sidebar,
    text="◈ NAVIGATION",
    font=("Arial", 12, "bold"),
    fg=CYAN,
    bg=HEADER
).pack(anchor="w", padx=25, pady=(25, 20))

for item in [
    "⌂   Dashboard",
    "◉   Skill Analyzer",
    "▣   Career Paths",
    "◇   Skill Intelligence",
    "⚙   AI Assistant"
]:
    tk.Label(
        sidebar,
        text=item,
        font=("Arial", 10, "bold"),
        fg=TEXT,
        bg=HEADER,
        anchor="w",
        padx=25,
        pady=12
    ).pack(fill="x")

tk.Label(
    sidebar,
    text="SYSTEM",
    font=("Arial", 9, "bold"),
    fg=MUTED,
    bg=HEADER
).pack(anchor="w", padx=25, pady=(35, 10))

tk.Label(
    sidebar,
    text="● Neural Engine Ready",
    font=("Arial", 9),
    fg=GREEN,
    bg=HEADER
).pack(anchor="w", padx=25)

# ============================================================
# CONTENT
# ============================================================

content = tk.Frame(body, bg=BG)
content.pack(side="left", fill="both", expand=True)

# ============================================================
# LEFT CONTROL PANEL
# ============================================================

left_panel = tk.Frame(content, bg=PANEL, width=330)
left_panel.pack(side="left", fill="y", padx=(15, 8), pady=15)
left_panel.pack_propagate(False)

tk.Label(
    left_panel,
    text="CAREER TARGET",
    font=("Arial", 10, "bold"),
    fg=MUTED,
    bg=PANEL
).pack(anchor="w", padx=20, pady=(20, 7))

career_var = tk.StringVar()

career_box = ttk.Combobox(
    left_panel,
    textvariable=career_var,
    values=list(CAREER_DATA.keys()),
    state="readonly",
    font=("Arial", 11)
)
career_box.pack(fill="x", padx=20, ipady=7)
career_box.set("Software Developer")

tk.Label(
    left_panel,
    text="YOUR CURRENT SKILLS",
    font=("Arial", 10, "bold"),
    fg=MUTED,
    bg=PANEL
).pack(anchor="w", padx=20, pady=(25, 7))

skills_entry = tk.Entry(
    left_panel,
    font=("Arial", 11),
    bg="#080B18",
    fg=WHITE,
    insertbackground=CYAN,
    relief="flat"
)
skills_entry.pack(fill="x", padx=20, ipady=10)

tk.Label(
    left_panel,
    text="Example: Python, SQL, Git",
    font=("Arial", 8),
    fg=MUTED,
    bg=PANEL
).pack(anchor="w", padx=20, pady=(5, 15))

# ============================================================
# DASHBOARD VARIABLES
# ============================================================

score_value = None
matched_value = None
missing_value = None
status_value = None

matched_text = None
missing_text = None
recommendation_text = None
graph_canvas = None

# ============================================================
# ANALYZE FUNCTION
# ============================================================

def analyze_skills():

    career = career_var.get()
    entered = skills_entry.get().strip()

    if not career:
        messagebox.showwarning("Career Required",
                               "Please select a career.")
        return

    if not entered:
        messagebox.showwarning("Skills Required",
                               "Please enter your current skills.")
        return

    user_skills = [
        skill.strip().lower()
        for skill in entered.split(",")
        if skill.strip()
    ]

    required_skills = [
        skill.lower()
        for skill in CAREER_DATA[career]["skills"]
    ]

    matched = [
        skill for skill in required_skills
        if skill in user_skills
    ]

    missing = [
        skill for skill in required_skills
        if skill not in user_skills
    ]

    score = round((len(matched) / len(required_skills)) * 100)

    score_value.config(text=f"{score}%")
    matched_value.config(text=str(len(matched)))
    missing_value.config(text=str(len(missing)))

    if score >= 80:
        status = "READY"
        status_color = GREEN
    elif score >= 50:
        status = "DEVELOPING"
        status_color = YELLOW
    else:
        status = "BEGINNER"
        status_color = RED

    status_value.config(text=status, fg=status_color)

    matched_text.delete("1.0", tk.END)

    if matched:
        matched_text.insert(
            tk.END,
            "\n".join("✓ " + x.title() for x in matched)
        )
    else:
        matched_text.insert(tk.END, "No matching skills found.")

    missing_text.delete("1.0", tk.END)

    if missing:
        missing_text.insert(
            tk.END,
            "\n".join("→ " + x.title() for x in missing)
        )
    else:
        missing_text.insert(
            tk.END,
            "No major skill gaps detected."
        )

    recommendation_text.delete("1.0", tk.END)

    if score >= 80:
        recommendation = (
            "Excellent progress!\n\n"
            "Your current skills strongly match "
            "the selected career path.\n\n"
            "Focus on advanced projects, certifications "
            "and practical experience."
        )
    elif score >= 50:
        recommendation = (
            "Good foundation!\n\n"
            "You already have several useful skills.\n\n"
            "Complete the missing skills and "
            "build practical projects."
        )
    else:
        recommendation = (
            "Skill development required.\n\n"
            "Start with the missing fundamental skills "
            "and follow the recommended learning path."
        )

    recommendation_text.insert(tk.END, recommendation)

    draw_graph(len(matched), len(missing))


# ============================================================
# CLEAR FUNCTION
# ============================================================

def clear_all():

    skills_entry.delete(0, tk.END)

    score_value.config(text="--")
    matched_value.config(text="--")
    missing_value.config(text="--")
    status_value.config(text="--", fg=WHITE)

    matched_text.delete("1.0", tk.END)
    missing_text.delete("1.0", tk.END)
    recommendation_text.delete("1.0", tk.END)

    matched_text.insert(
        tk.END,
        "Analysis results will appear here."
    )

    missing_text.insert(
        tk.END,
        "Skill gaps will appear here."
    )

    recommendation_text.insert(
        tk.END,
        "AI recommendations will appear here."
    )

    graph_canvas.delete("all")

# ============================================================
# BUTTONS
# ============================================================

tk.Button(
    left_panel,
    text="⚡  ANALYZE SKILLS",
    command=analyze_skills,
    font=("Arial", 11, "bold"),
    bg=CYAN,
    fg="#050814",
    activebackground=PURPLE,
    activeforeground=WHITE,
    relief="flat",
    cursor="hand2"
).pack(fill="x", padx=20, pady=(10, 8), ipady=10)

tk.Button(
    left_panel,
    text="↻  CLEAR",
    command=clear_all,
    font=("Arial", 10, "bold"),
    bg=PANEL2,
    fg=TEXT,
    activebackground="#20284A",
    activeforeground=WHITE,
    relief="flat",
    cursor="hand2"
).pack(fill="x", padx=20, pady=5, ipady=8)

# ============================================================
# RIGHT DASHBOARD
# ============================================================

dashboard = tk.Frame(content, bg=BG)
dashboard.pack(side="left", fill="both", expand=True,
               padx=(8, 15), pady=15)

# ============================================================
# ROBOT AREA
# ============================================================

robot_frame = tk.Frame(
    dashboard,
    bg=HEADER,
    height=180
)
robot_frame.pack(fill="x", pady=(0, 10))
robot_frame.pack_propagate(False)

robot_canvas = tk.Canvas(
    robot_frame,
    bg=HEADER,
    highlightthickness=0
)
robot_canvas.pack(fill="both", expand=True)

# ============================================================
# PARTICLES
# ============================================================

particles = []

for _ in range(45):

    x = random.randint(0, 900)
    y = random.randint(0, 180)

    particle = robot_canvas.create_oval(
        x, y, x + 2, y + 2,
        fill=CYAN,
        outline=""
    )

    particles.append({
        "id": particle,
        "x": x,
        "y": y,
        "speed": random.uniform(0.2, 0.8)
    })

def animate_particles():

    width = robot_canvas.winfo_width()

    for p in particles:

        p["x"] -= p["speed"]

        if p["x"] < 0:
            p["x"] = width

        robot_canvas.coords(
            p["id"],
            p["x"],
            p["y"],
            p["x"] + 2,
            p["y"] + 2
        )

    root.after(40, animate_particles)

# ============================================================
# FIXED ROUNDED RECTANGLE FUNCTION
# ============================================================

def create_rounded_rectangle(
    canvas,
    x1,
    y1,
    x2,
    y2,
    radius=10,
    **kwargs
):

    points = [
        x1 + radius, y1,
        x2 - radius, y1,
        x2, y1 + radius,
        x2, y2 - radius,
        x2 - radius, y2,
        x1 + radius, y2,
        x1, y2 - radius,
        x1, y1 + radius
    ]

    return canvas.create_polygon(
        points,
        smooth=True,
        **kwargs
    )

# ============================================================
# ROBOT
# ============================================================

robot_parts = []

def draw_robot():

    for part in robot_parts:
        robot_canvas.delete(part)

    robot_parts.clear()

    cx = robot_canvas.winfo_width() // 2
    cy = 88

    # Outer holographic rings
    robot_parts.append(
        robot_canvas.create_oval(
            cx - 75, cy - 75,
            cx + 75, cy + 75,
            outline="#14294A",
            width=2
        )
    )

    robot_parts.append(
        robot_canvas.create_oval(
            cx - 62, cy - 62,
            cx + 62, cy + 62,
            outline="#10203A",
            width=2
        )
    )

    # Robot head
    robot_parts.append(
        create_rounded_rectangle(
            robot_canvas,
            cx - 45,
            cy - 45,
            cx + 45,
            cy + 15,
            radius=12,
            outline=CYAN,
            width=2,
            fill="#101A30"
        )
    )

    # Head antenna
    robot_parts.append(
        robot_canvas.create_line(
            cx,
            cy - 45,
            cx,
            cy - 60,
            fill=PURPLE,
            width=2
        )
    )

    robot_parts.append(
        robot_canvas.create_oval(
            cx - 4,
            cy - 65,
            cx + 4,
            cy - 57,
            fill=PINK,
            outline=""
        )
    )

    # Eyes
    robot_parts.append(
        robot_canvas.create_oval(
            cx - 27,
            cy - 22,
            cx - 12,
            cy - 7,
            fill=CYAN,
            outline=""
        )
    )

    robot_parts.append(
        robot_canvas.create_oval(
            cx + 12,
            cy - 22,
            cx + 27,
            cy - 7,
            fill=CYAN,
            outline=""
        )
    )

    # Mouth
    robot_parts.append(
        robot_canvas.create_line(
            cx - 12,
            cy + 2,
            cx + 12,
            cy + 2,
            fill=PURPLE,
            width=2
        )
    )

    # Body
    robot_parts.append(
        robot_canvas.create_rectangle(
            cx - 38,
            cy + 18,
            cx + 38,
            cy + 65,
            outline=PURPLE,
            width=2,
            fill="#111A32"
        )
    )

    # Left arm
    robot_parts.append(
        robot_canvas.create_line(
            cx - 38,
            cy + 25,
            cx - 65,
            cy + 48,
            fill=CYAN,
            width=4
        )
    )

    # Right arm
    robot_parts.append(
        robot_canvas.create_line(
            cx + 38,
            cy + 25,
            cx + 65,
            cy + 48,
            fill=CYAN,
            width=4
        )
    )

    # Chest core
    robot_parts.append(
        robot_canvas.create_oval(
            cx - 12,
            cy + 30,
            cx + 12,
            cy + 54,
            fill=CYAN,
            outline=""
        )
    )

    # Chest core ring
    robot_parts.append(
        robot_canvas.create_oval(
            cx - 18,
            cy + 24,
            cx + 18,
            cy + 60,
            outline=CYAN,
            width=1
        )
    )

def robot_animation():
    draw_robot()
    root.after(700, robot_animation)

# ============================================================
# STAT CARDS
# ============================================================

stats = tk.Frame(dashboard, bg=BG)
stats.pack(fill="x", pady=(0, 10))

def create_stat(parent, title):

    frame = tk.Frame(
        parent,
        bg=PANEL,
        height=85
    )

    frame.pack(
        side="left",
        fill="both",
        expand=True,
        padx=4
    )

    tk.Label(
        frame,
        text=title,
        font=("Arial", 8, "bold"),
        fg=MUTED,
        bg=PANEL
    ).pack(anchor="w", padx=12, pady=(10, 2))

    value = tk.Label(
        frame,
        text="--",
        font=("Arial", 19, "bold"),
        fg=CYAN,
        bg=PANEL
    )

    value.pack(anchor="w", padx=12)

    return value

score_value = create_stat(stats, "AI MATCH SCORE")
matched_value = create_stat(stats, "MATCHED SKILLS")
missing_value = create_stat(stats, "SKILL GAPS")
status_value = create_stat(stats, "READINESS STATUS")

# ============================================================
# LOWER DASHBOARD
# ============================================================

lower = tk.Frame(dashboard, bg=BG)
lower.pack(fill="both", expand=True)

# ============================================================
# GRAPH PANEL
# ============================================================

graph_panel = tk.Frame(
    lower,
    bg=PANEL,
    width=450
)
graph_panel.pack(
    side="left",
    fill="both",
    expand=True,
    padx=(0, 5)
)

tk.Label(
    graph_panel,
    text="SKILL INTELLIGENCE",
    font=("Arial", 11, "bold"),
    fg=CYAN,
    bg=PANEL
).pack(anchor="w", padx=15, pady=12)

graph_canvas = tk.Canvas(
    graph_panel,
    bg=HEADER,
    highlightthickness=0
)
graph_canvas.pack(
    fill="both",
    expand=True,
    padx=15,
    pady=(0, 15)
)

def draw_graph(matched, missing):

    graph_canvas.delete("all")

    width = graph_canvas.winfo_width()
    height = graph_canvas.winfo_height()

    if width < 100:
        width = 400

    if height < 100:
        height = 250

    total = matched + missing

    if total == 0:
        return

    cx = width // 2
    cy = height // 2
    radius = min(width, height) // 3

    graph_canvas.create_oval(
        cx - radius,
        cy - radius,
        cx + radius,
        cy + radius,
        outline="#26314F",
        width=20
    )

    matched_angle = (matched / total) * 360

    graph_canvas.create_arc(
        cx - radius,
        cy - radius,
        cx + radius,
        cy + radius,
        start=90,
        extent=-matched_angle,
        style="arc",
        outline=CYAN,
        width=20
    )

    score = round((matched / total) * 100)

    graph_canvas.create_text(
        cx,
        cy - 5,
        text=f"{score}%",
        fill=WHITE,
        font=("Arial", 24, "bold")
    )

    graph_canvas.create_text(
        cx,
        cy + 25,
        text="READINESS",
        fill=MUTED,
        font=("Arial", 8, "bold")
    )

# ============================================================
# RIGHT INFORMATION PANEL
# ============================================================

right_panel = tk.Frame(lower, bg=BG)
right_panel.pack(
    side="left",
    fill="both",
    expand=True,
    padx=(5, 0)
)

# ============================================================
# MATCHED SKILLS
# ============================================================

matched_frame = tk.Frame(right_panel, bg=PANEL)
matched_frame.pack(fill="both", expand=True, pady=(0, 5))

tk.Label(
    matched_frame,
    text="✓  MATCHED SKILLS",
    font=("Arial", 10, "bold"),
    fg=GREEN,
    bg=PANEL
).pack(anchor="w", padx=12, pady=8)

matched_text = tk.Text(
    matched_frame,
    bg=PANEL,
    fg=TEXT,
    font=("Arial", 9),
    relief="flat",
    height=5
)
matched_text.pack(fill="both", expand=True, padx=12, pady=(0, 10))
matched_text.insert(tk.END, "Analysis results will appear here.")

# ============================================================
# MISSING SKILLS
# ============================================================

missing_frame = tk.Frame(right_panel, bg=PANEL)
missing_frame.pack(fill="both", expand=True, pady=5)

tk.Label(
    missing_frame,
    text="→  SKILL GAPS",
    font=("Arial", 10, "bold"),
    fg=RED,
    bg=PANEL
).pack(anchor="w", padx=12, pady=8)

missing_text = tk.Text(
    missing_frame,
    bg=PANEL,
    fg=TEXT,
    font=("Arial", 9),
    relief="flat",
    height=5
)
missing_text.pack(fill="both", expand=True, padx=12, pady=(0, 10))
missing_text.insert(tk.END, "Skill gaps will appear here.")

# ============================================================
# AI RECOMMENDATION
# ============================================================

recommend_frame = tk.Frame(right_panel, bg=PANEL)
recommend_frame.pack(fill="both", expand=True, pady=(5, 0))

tk.Label(
    recommend_frame,
    text="✦  AI RECOMMENDATION ENGINE",
    font=("Arial", 10, "bold"),
    fg=PURPLE,
    bg=PANEL
).pack(anchor="w", padx=12, pady=8)

recommendation_text = tk.Text(
    recommend_frame,
    bg=PANEL,
    fg=TEXT,
    font=("Arial", 9),
    relief="flat",
    height=6
)
recommendation_text.pack(
    fill="both",
    expand=True,
    padx=12,
    pady=(0, 10)
)
recommendation_text.insert(
    tk.END,
    "AI recommendations will appear here."
)

# ============================================================
# COURSES
# ============================================================

def show_courses():

    career = career_var.get()

    if not career:
        messagebox.showwarning(
            "Select Career",
            "Please select a career first."
        )
        return

    window = tk.Toplevel(root)
    window.title("Recommended Courses")
    window.geometry("550x500")
    window.configure(bg=BG)

    tk.Label(
        window,
        text="RECOMMENDED LEARNING PATH",
        font=("Arial", 16, "bold"),
        fg=CYAN,
        bg=BG
    ).pack(pady=20)

    for course in CAREER_DATA[career]["courses"]:

        tk.Label(
            window,
            text="◆  " + course,
            font=("Arial", 11),
            fg=TEXT,
            bg=BG,
            anchor="w"
        ).pack(fill="x", padx=50, pady=7)

# ============================================================
# ROADMAP
# ============================================================

def show_roadmap():

    career = career_var.get()

    if not career:
        messagebox.showwarning(
            "Select Career",
            "Please select a career first."
        )
        return

    window = tk.Toplevel(root)
    window.title("Career Roadmap")
    window.geometry("600x550")
    window.configure(bg=BG)

    tk.Label(
        window,
        text=f"{career.upper()} ROADMAP",
        font=("Arial", 18, "bold"),
        fg=CYAN,
        bg=BG
    ).pack(pady=20)

    steps = [
        "1. Learn the fundamentals",
        "2. Build technical skills",
        "3. Practice problem solving",
        "4. Complete practical projects",
        "5. Learn industry tools",
        "6. Earn relevant certifications",
        "7. Build GitHub portfolio",
        "8. Apply for internships",
        "9. Prepare for interviews"
    ]

    for step in steps:

        tk.Label(
            window,
            text=step,
            font=("Arial", 11),
            fg=TEXT,
            bg=BG,
            anchor="w"
        ).pack(fill="x", padx=70, pady=8)

# ============================================================
# AI ASSISTANT
# ============================================================

def show_assistant():

    window = tk.Toplevel(root)
    window.title("CareerCompass AI Assistant")
    window.geometry("600x600")
    window.configure(bg=BG)

    tk.Label(
        window,
        text="✦ CAREER AI ASSISTANT",
        font=("Arial", 18, "bold"),
        fg=CYAN,
        bg=BG
    ).pack(pady=15)

    chat = tk.Text(
        window,
        bg=PANEL,
        fg=TEXT,
        font=("Arial", 10),
        relief="flat"
    )
    chat.pack(fill="both", expand=True, padx=20, pady=10)

    chat.insert(
        tk.END,
        "AI: Hello! I am CareerCompass AI.\n\n"
        "Ask me about:\n"
        "• Career skills\n"
        "• Skill gaps\n"
        "• Learning roadmap\n"
        "• Courses\n"
        "• Cyber Security\n"
        "• Projects\n\n"
    )

    entry = tk.Entry(
        window,
        bg=HEADER,
        fg=WHITE,
        insertbackground=CYAN,
        font=("Arial", 10)
    )
    entry.pack(fill="x", padx=20, pady=10, ipady=8)

    def reply():

        question = entry.get().lower()

        if "cyber" in question:
            answer = (
                "For Cyber Security, focus on Networking, "
                "Linux, Python, Cyber Security, "
                "Cryptography and Ethical Hacking."
            )

        elif "python" in question:
            answer = (
                "Python is useful for software development, "
                "data analysis, AI/ML and cyber security."
            )

        elif "career" in question:
            answer = (
                "Choose a career based on your interests "
                "and compare your current skills with "
                "the required skills."
            )

        elif "project" in question:
            answer = (
                "Build small practical projects and upload "
                "them to GitHub to create your portfolio."
            )

        else:
            answer = (
                "Keep learning step by step. Focus on the "
                "missing skills shown by CareerCompass."
            )

        chat.insert(tk.END, f"\nYOU: {entry.get()}\n")
        chat.insert(tk.END, f"AI: {answer}\n")
        entry.delete(0, tk.END)

    tk.Button(
        window,
        text="SEND",
        command=reply,
        bg=CYAN,
        fg="#050814",
        font=("Arial", 10, "bold"),
        relief="flat"
    ).pack(pady=(0, 20), ipadx=30, ipady=6)

# ============================================================
# BOTTOM BUTTONS
# ============================================================

button_bar = tk.Frame(dashboard, bg=BG)
button_bar.pack(fill="x", pady=(8, 0))

def dashboard_button(text, command):

    return tk.Button(
        button_bar,
        text=text,
        command=command,
        font=("Arial", 9, "bold"),
        bg=PANEL2,
        fg=TEXT,
        activebackground=PURPLE,
        activeforeground=WHITE,
        relief="flat",
        cursor="hand2"
    )

dashboard_button(
    "▣ COURSES",
    show_courses
).pack(side="left", padx=4, ipadx=12, ipady=7)

dashboard_button(
    "◇ ROADMAP",
    show_roadmap
).pack(side="left", padx=4, ipadx=12, ipady=7)

dashboard_button(
    "✦ AI ASSISTANT",
    show_assistant
).pack(side="left", padx=4, ipadx=12, ipady=7)

# ============================================================
# START
# ============================================================

root.after(100, animate_particles)
root.after(100, robot_animation)

root.mainloop()
```
